# Test VisDoMRAG.extract_text_from_pdf (DotsOCR)

This notebook tests the current implementation of `VisDoMRAG.extract_text_from_pdf` in `visdom.py`,
using a lightweight instance of `VisDoMRAG` (bypassing full LLM initialization).


```python
import os
from visdom import VisDoMRAG

# Set the PDF path you want to test
pdf_path = '/home/zechuan/m3docrag/contents/2024_Tencent_ESG.pdf'  # TODO: change if needed
assert os.path.exists(pdf_path), f'PDF not found: {pdf_path}'

# Create a lightweight VisDoMRAG instance without running full __init__
vis = VisDoMRAG.__new__(VisDoMRAG)
```

```python
# DotsOCR-related config can be adjusted here if needed
vis.config = {
    'dots_prompt_mode': 'prompt_layout_all_en',
    'dots_num_thread': 64,
}

pages = vis.extract_text_from_pdf(pdf_path)

print(f'Total pages extracted: {len(pages)}')
for i, page in enumerate(pages[:3]):
    print('=' * 80)
    print(f'Page {i+1} preview:')
    print(page[:1000])  # print first 1000 chars of each page
```
